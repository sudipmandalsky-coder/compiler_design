```python
import numpy as np

def create_matrix():
    # Example of creating a matrix from a string
    matrix_str = """1 2 3
                   4 5 6
                   7 8 9"""
    matrix = np.array(matrix_str.split(), dtype=float).reshape(3, 3)
    return matrix

def main():
    # Step 1: Create two matrices
    matrix1 = create_matrix()
    print("Matrix 1:\n", matrix1)

    matrix2 = create_matrix()
    print("Matrix 2:\n", matrix2)

    # Step 2: Matrix Addition
    result_add = matrix1 + matrix2
    print("Matrix Addition:\n", result_add)

    # Step 3: Matrix Multiplication
    result_mul = np.dot(matrix1, matrix2)
    print("Matrix Multiplication:\n", result_mul)

    # Step 4: Inverse of Matrix
    if np.linalg.det(matrix1) != 0:
        result_inv = np.linalg.inv(matrix1)
        print("Inverse of Matrix:\n", result_inv)
    else:
        print("The matrix is not invertible.")

if __name__ == "__main__":
    main()
```
